### Calculating the Corrected Forbes Similarity Coefficient Between Sets 

[Forbes Similarity Coefficient]("https://bio.mq.edu.au/~jalroy/Forbes.html") 
- F' = a (n + sqrt n)/[a (n + sqrt n) + 3/2 b c]

Where N = the number of different things, a = the number found in both lists, b = the number found only in the first list, c = the number found only in the second, and d = the number found in neither one. By definition, N = a + b + c + d.

```python
sonis_dict = read_json(./gene_set_results/sonis_dict.json)
```

```python
def forbes_similarity(dict1,dict2):
    sim_list = []
    for (x,l1) in dict1.items():
        for (y,l2) in dict2.items():
            a = len([t for t in l1 if t in l2])
            b = len([t for t in l1 if not t in l2])
            c = len([t for t in l2 if not t in l1])
            n = a + b + c
            F = a * (n + sqrt(n)) / (a * (n + sqrt(n)) + (3/2) *b*c)
            sim_list.append((x,y,F))
    
    #make pivot dataframe
    p1 = []
    p2 = []
    forbes = []
    for f in sim_list:
        x,y,z = f
        p1.append(x)
        p2.append(y)
        forbes.append(z)
    df_2 = pd.DataFrame({"path1":p1, "path2":p2, "Forbes":forbes})
    df_pivot = pd.pivot_table(df_2,index=["path1"],columns=["path2"],values=["Forbes"])
    
    #plot heat map
    plt.figure(figsize=(20,20))
    fig = sns.heatmap(df_pivot, xticklabels=True, yticklabels=True, cmap="Blues")
    fig.set(xlabel="pathway_2", ylabel="pathway_1")
    plt.savefig("graphs\heatmap")
    
    return df_2                
```

```python
def jaccard_index(dict1, dict2):
    sim_list = []
    for (a , x) in dict1.items():
        for (b , y) in dict2.items():
            overlap = len(set(x).intersection(set(y)))
            combine = len(set(x).union(set(y)))
            j = overlap/combine
            sim_list.append((a,b,j))
            sim_list.append((b,a,j))
    return sim_list
```

```python
external_overlap = jaccard_index(sonis_dict,string_candidate_gene_dict) + jaccard_index(sonis_dict, pk_dict) + jaccard_index(sonis_dict, pd_dict) + jaccard_index(string_candidate_gene_dict, pd_dict) + jaccard_index(string_candidate_gene_dict, pk_dict) + jaccard_index(pk_dict, pd_dict)
```

```python
def internal_overlap(dict):
    inter_list = []
    genes = list(dict.values())
    keys = list(dict.keys())
    for x in range(0,len(keys)):
        for y in range(0,len(keys)):
            j2 = len(set(genes[x]).intersection(set(genes[y])))/len(set(genes[x]).union(set(genes[y])))
            inter_list.append((keys[x], keys[y], j2))
            inter_list.append((keys[y], keys[x], j2))
            y += 1
        x += 1
    return inter_list
```

```python
sonis_internal = internal_overlap(sonis_dict)
string_internal = internal_overlap(string_candidate_gene_dict)
internal_overlap = sonis_internal + string_internal
```

```python
internal_overlap = sonis_internal + string_internal
```

```python
combined_list =  external_overlap + internal_overlap
len(combined_list)
```

```python
p1 = []
p2 = []
j = []
for x in combined_list:
    a,b,c = x
    p1.append(a)
    p2.append(b)
    j.append(c)

df = pd.DataFrame({"path1":p1, "path2":p2, "jaccard":j})
```

```python
import gseapy as gp
def get_enrichr_txt(input_list, library):
    enr = gp.enrichr(gene_list=input_list,
                     gene_sets= library,
                     organism='Human', # don't forget to set organism to the one you desired! e.g. Yeast
                     description="sonis_pathways",
                     no_plot=True,
                     cutoff = 0.05 #  Show enriched terms which Adjusted P-value < cutoff.
                    )
    df = enr.results
    return df
```

```python
def enrichr(dict,library):
    df = pd.DataFrame(columns=["Gene_set", "Term","Overlap","original_size"])
    for sublist in dict.values():
        df_enr = get_enrichr_txt(sublist, library).sort_values(["Combined Score"], ascending=False).iloc[0:1]
        df_enr["original_size"] = len(sublist)
        df_enr = df_enr.drop([ "Adjusted P-value", "Old P-value", "Old Adjusted P-value", "Odds Ratio", "P-value", "Combined Score","Genes"], axis=1 )
        df = pd.concat([df,df_enr],ignore_index=True)
   
    return df
```

```python
def gs_extraction(db_path,name):
    with open(db_path, "r")as h:   
        for l in h.readlines():
            a = l.split("\t")
            if a[0] == name:
                 gs = a[2:-2]
        return gs
```

```python
import mygene

def get_id(gene_list):
    mg = mygene.MyGeneInfo()
    out = mg.querymany(gene_list, scopes='symbol', fields='entrezgene', species='human')
    entrezid_list = []
    problem_list = []
#     print(out)
    for gene in out:
        try:
            entrezid_list.append(gene["entrezgene"])
        except:
            problem_list.append(gene["query"])
            
    print("notfound", set(problem_list))
    print (entrezid_list)
    return entrezid_list  
```
