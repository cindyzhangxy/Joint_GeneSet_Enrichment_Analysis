## Using Enrichr for Gene Set reconcilliation 

-  'KEGG_2021_Human','Reactome_2016', 'BioPlanet_2019', 'WikiPathways_2019_Human', "Human_Phenotype_Ontology", and "GO_Biological_Process_2021"

```python
library = ['Reactome_2016', 'KEGG_2021_Human','BioPlanet_2019','WikiPathways_2019_Human', 'Human_Phenotype_Ontology', 'GO_Biological_Process_2021']
```

```python
import gseapy as gp
def get_enrichr_txt(input_list, library):
    enr = gp.enrichr(gene_list=input_list,
                     gene_sets= library,
                     organism='Human', # don't forget to set organism to the one you desired! e.g. Yeast
                     description="sonis_pathways",
                     no_plot=True,
                     cutoff = 0.05 #  Show enriched terms which Adjusted P-value < cutoff.
                    )
    df = enr.results
    return df
```

```python
df_list = []
for sublist in dict.values():    
    df_enr = get_enrichr_txt(sublist, library).sort_values(["Combined Score"], ascending=False).iloc[0:1]
    df_enr["original size"] = len(sublist)
    new_df = df_enr.drop(["Adjusted P-value", "Old P-value", "Old Adjusted P-value", "Odds Ratio", "P-value", "Combined Score"], axis=1 )
    df_list.append(new_df)
    
summary_df = pd.concat(df_list)
```

```python
summary_df.to_excel("./gene_set_results/enrichr_sonis.xlsx")
```

```python
summary_df["original_set"] = dict.keys()
summary_df.rename(columns = {"Gene_set":"library", "Term":"term", "Overlap":"overlap"}, inplace=True)
summary_df.loc[:,["library","term","original_set","overlap","original size"]]
```

## Add Extra Genes from Enrichr

```python
def extra_genes(lib, enr_path, s_path):
    with open (f"Enrichr\{lib}") as f:
        for line in f.readlines():
            line = line.split("\t")
            if line[0] == enr_path:
                enr_list = line[2:-1]
                ext = set(dict[s_path]).union(set(enr_list))
                print(enr_path," added : ", len(ext)-len(set(dict[s_path])))
                ext = [x for x in ext]
                return ext
```

```python
new_toll = extra_genes("BioPlanet_2019.txt","Toll-like receptor signaling pathway regulation","KEGG_TOLL_LIKE_RECEPTOR_SIGNALING_PATHWAY")
new_int = extra_genes("BioPlanet_2019.txt","Platelet aggregation (plug formation)","REACTOME_INTEGRIN_SIGNALING")
new_wnt = extra_genes("BioPlanet_2019.txt","Inactivation of GSK3 by Akt causes accumulation of beta-catenin in alveolar macrophages","WP_WNTBETACATENIN_SIGNALING_PATHWAY_IN_LEUKEMIA")
new_jnk = extra_genes("GO_Biological_Process_2021.txt","JNK cascade (GO:0007254)","REACTOME_JNK_C_JUN_KINASES_PHOSPHORYLATION_AND_ACTIVATION_MEDIATED_BY_ACTIVATED_HUMAN_TAK1")
```

- Toll-like receptor signaling pathway regulation  added :  40
- Platelet aggregation (plug formation)  added :  10
- Inactivation of GSK3 by Akt causes accumulation of beta-catenin in alveolar macrophages  added :  20
- JNK cascade (GO:0007254)  added :  19

```python
#update sonis pathways with the Enrichr added gene sets 
dict["KEGG_TOLL_LIKE_RECEPTOR_SIGNALING_PATHWAY"] = new_toll
dict["REACTOME_JNK_C_JUN_KINASES_PHOSPHORYLATION_AND_ACTIVATION_MEDIATED_BY_ACTIVATED_HUMAN_TAK1"] = new_int
dict["WP_WNTBETACATENIN_SIGNALING_PATHWAY_IN_LEUKEMIA"] = new_wnt
dict["REACTOME_JNK_C_JUN_KINASES_PHOSPHORYLATION_AND_ACTIVATION_MEDIATED_BY_ACTIVATED_HUMAN_TAK1"] = new_jnk 
```

```python
# store output to gene_set_results folder
with open("./gene_set_results/sonis_dict.json", 'w') as fp:
    json.dump(dict, fp)
```
