## Methotrexate-specific pharmacodynamics and pharmacokinetics gene sets were curated from [PharmGKB](https://www.pharmgkb.org/)  

```python
import json
```

```python
# Pharmacokinetics gene set
pk_genes = ["ABCB1", 'ABCC1', 'ABCC2', 'ABCC3', 'ABCG2', 'FOLR1', 'SLC19A1', 'SLC22A6', 'SLC22A8', 'SLC46A1', 'SLCO1A2', 'SLCO1B1', 'SLCO1B3']
```

```python
# Pharmacodynamics gene set 
pd_genes = ['GGH','FGPS','ADA', 'ATIC', 'CBS', 'DHFR', 'GART', 'MTHFD1', 'MTHFR', 'MTHFS', 'MTR', 'MTRR', 'NT5E', 'PPAT', 'SHMT1', 'TYMS']
```

```python
with open("./data/pk_pd_list.json", "w") as g:
    json.dump(pd_genes+pk_genes, g) 
```

```python
pk_dict = {"pk":pk_genes}
pd_dict = {"pd":pd_genes}
with open ("./gene_set_results/pk_dict.json", "w") as f:
    json.dump(pk_dict, f)
with open("./gene_set_results/pd_dict.json", "w") as g:
    json.dump(pd_dict, g)
```
