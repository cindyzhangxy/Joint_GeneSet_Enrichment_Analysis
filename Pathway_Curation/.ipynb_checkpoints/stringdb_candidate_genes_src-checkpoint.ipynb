### 50 Functional Gene Neightbours for each of the above Candidate Genes

```python
def find_neighbours(gene, nodes): 

    string_api_url = "https://version-11-5.string-db.org/api"
    output_format = "tsv-no-header"
    method = "interaction_partners"

    ##
    ## Construct the request
    ##

    request_url = "/".join([string_api_url, output_format, method])

    ##
    ## Set parameters
    ##
    my_genes = [gene]
    
    params = {

        "identifiers" : "%0d".join(my_genes), # your protein
        "species" : 9606, # species NCBI identifier 
        "limit" : nodes ,
        "network type" : "functional", 

    }

    ##
    ## Call STRING
    ##

    response = requests.post(request_url, data=params)

    ##
    ## Read and parse the results
    ##
    list = [gene]
    for line in response.text.strip().split("\n"):

        l = line.strip().split("\t")
        query_ensp = l[0]
        query_name = l[2]
        partner_ensp = l[1]
        partner_name = l[3] 
        combined_score = l[5]
        list.append(partner_name)

    return(list)
```

```python
string_candidate_gene_dict={}
for gene in CANDIDATE:
    string_candidate_gene_dict[gene] = find_neighbours(gene,50)
```

### Calculating the Corrected Forbes Similarity Coefficient Between Sets 

[Forbes Similarity Coefficient]("https://bio.mq.edu.au/~jalroy/Forbes.html") 
- F' = a (n + sqrt n)/[a (n + sqrt n) + 3/2 b c]

Where N = the number of different things, a = the number found in both lists, b = the number found only in the first list, c = the number found only in the second, and d = the number found in neither one. By definition, N = a + b + c + d.

```python
def forbes_similarity(dict1,dict2):
    sim_list = []
    for (x,l1) in dict1.items():
        for (y,l2) in dict2.items():
            a = len([t for t in l1 if t in l2])
            b = len([t for t in l1 if not t in l2])
            c = len([t for t in l2 if not t in l1])
            n = a + b + c
            m = n + np.sqrt(n)
            F = a * m / (a *m + (3/2)*b*c)  
            sim_list.append((x,y,F))
    
    #make pivot dataframe
    p1 = []
    p2 = []
    forbes = []
    for f in sim_list:
        x,y,z = f
        p1.append(x)
        p2.append(y)
        forbes.append(z)
    df_2 = pd.DataFrame({"path1":p1, "path2":p2, "Forbes":forbes})
    df_pivot = pd.pivot_table(df_2,index=["path1"],columns=["path2"],values=["Forbes"])
    
    #plot heat map
    plt.figure(figsize=(20,20))
    fig = sns.heatmap(df_pivot, xticklabels=True, yticklabels=True, cmap="Blues_r")
    fig.set(xlabel="pathway_2", ylabel="pathway_1")
    plt.savefig("./graphs/heatmap")
    
    return df_2
```

## Enrichr Top Theme from MSigDB

```python
library_1 = ["MSigDB_Hallmark_2020"]
```

```python
import gseapy as gp
def get_enrichr_txt(input_list, library):
    enr = gp.enrichr(gene_list=input_list,
                     gene_sets= library,
                     organism='Human', # don't forget to set organism to the one you desired! e.g. Yeast
                     description="sonis_pathways",
                     no_plot=True,
                     cutoff = 0.05 #  Show enriched terms which Adjusted P-value < cutoff.
                    )
    df = enr.results
    return df

def enrichr(dict):
    df_list = []
    for sublist in dict.values():    
        df_enr = get_enrichr_txt(sublist, library_1).sort_values(["Combined Score"], ascending=False).iloc[0:1]
        df_enr["original size"] = len(sublist)
        new_df = df_enr.drop(["Adjusted P-value", "Old P-value", "Old Adjusted P-value", "Odds Ratio", "P-value","Combined Score"], axis=1 )
        df_list.append(new_df)

    summary_df = pd.concat(df_list)
    return summary_df
```

- How frequently does each gene occur in all novo pathways?

```python
def count_genes(list):
    frequency_counter = {}
    for gene in list:
        if gene in frequency_counter:
            frequency_counter[gene] += 1
        else:
            frequency_counter[gene] = 1
    with open ("stringdb_gene_frequency", "w") as f:
        print(frequency_counter, file=f) 
```

- <span style='color:Blue'>  How connected is each gene with all other genes?  

```python
def interactions(genes):
    string_api_url = "https://version-11-5.string-db.org/api"
    output_format = "json"
    method = "interaction_partners"
    
    new_dict = {}
    for gene in genes:
        request_url = "/".join([string_api_url, output_format, method])

        input_gene = [gene]

        params = {

            "identifiers": "%0d".join(input_gene), # your proteins
            "species": 9606,
            "limit": 9000,   
            "network_type": "functional"

        }

        response = requests.post(request_url, data=params)
        data = json.loads(response.text)
        
        new_dict[gene] = len(data)
        
    with open ("stringdb_interactions", "w") as f:
        print(new_dict, file=f)
    
    df = pd.DataFrame(new_dict.items(), columns=["gene", "interactions"])
    return df
```

- <span style='color:blue'> The below method retrieve functional annotation using stringDB</span>

```python
def functional_annotation(genes):
    
    string_api_url = "https://version-11-5.string-db.org/api"
    output_format = "json"
    method = "functional_annotation"

    request_url = "/".join([string_api_url, output_format, method])

    my_genes = genes

    params = {

        "identifiers" : "%0d".join(my_genes), # your protein
        "species" : 9606, # species NCBI identifier 
        "allow_pubmed" : 0

    }

    response = requests.post(request_url, data=params)

    data = json.loads(response.text)

    for row in data:

        term = row["term"]
        preferred_names = ",".join(row["preferredNames"])
        description = row["description"]
        category = row["category"]
        number_of_genes = row["number_of_genes"]
        ratio_in_set = row["ratio_in_set"]
        
        if term == "WikiPathways" :
            print("\t".join([category, str(ratio_in_set), description]))
            print()
```

```python
for genes in string_candidate_gene_dict: 
    functional_annotation(genes) 
```

```python
def strigndb_enrichment(genes):
    
    string_api_url = "https://version-11-5.string-db.org/api"
    output_format = "json"
    method = "enrichment"

    request_url = "/".join([string_api_url, output_format, method])

    my_genes = genes

    params = {

        "identifiers" : "%0d".join(my_genes), # your protein
        "species" : 9606, # species NCBI identifier 


    }

    response = requests.post(request_url, data=params)
    
    data = json.loads(response.text)

    for row in data:

        term = row["term"]
        preferred_names = ",".join(row["preferredNames"])
        fdr = float(row["fdr"])
        description = row["description"]
        category = row["category"]
        number_of_genes = row["number_of_genes"]

        if category == "Process" and fdr < 0.01:

            # print significant GO Process annotations
#             print(str(fdr))
            
            print("\t".join([preferred_names, str(number_of_genes), description, category]))
```
