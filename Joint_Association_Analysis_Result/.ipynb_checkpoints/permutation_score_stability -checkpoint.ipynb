```python
import pandas as pd
import numpy as np
import seaborn as sns 
```

```python
df_perm = pd.read_csv("./jag.merged.P1.perm", sep="\t")
```

```python
df_real =  pd.read_csv("./jag.merged.P1.empp",sep="\t")
```

```python
# store permutation result 

def perm(j):
    perm = []
    for i in range(5800,10000):
        r = pd.Series.tolist(df_perm.iloc[0:i+1,j])
        perm.append(r)

    filter = []

    for r in perm:
        filter.append(len([num for num in r if num > df_real["sumlogReal"][j]])/len(r))

    return filter
```

```python
perm_list = []
for j in range(0,17):
    perm_list.append(perm(j))
```
